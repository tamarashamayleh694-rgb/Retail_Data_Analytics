```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
sales = pd.read_csv("sales data-set.csv")
features = pd.read_csv("Features data set.csv")
stores = pd.read_csv("stores data-set.csv")

print("Sales Dataset Shape:", sales.shape)
print("Features Dataset Shape:", features.shape)
print("Stores Dataset Shape:", stores.shape)

print("\nSales Dataset:")
display(sales.head())

print("\nFeatures Dataset:")
display(features.head())

print("\nStores Dataset:")
display(stores.head())
```

```python
print("===== SALES DATASET =====")
sales.info()

print("\n===== FEATURES DATASET =====")
features.info()

print("\n===== STORES DATASET =====")
stores.info()
```

```python
print("Missing Values in Sales:")
print(sales.isnull().sum())

print("\nMissing Values in Features:")
print(features.isnull().sum())

print("\nMissing Values in Stores:")
print(stores.isnull().sum())
```

```python
print("Sales Summary:")
display(sales.describe())

print("Features Summary:")
display(features.describe())

print("Stores Summary:")
display(stores.describe())
```

```python
# Convert Date columns to datetime

sales['Date'] = pd.to_datetime(sales['Date'], format='%d/%m/%Y')
features['Date'] = pd.to_datetime(features['Date'], format='%d/%m/%Y')

# Check duplicate rows

print("Duplicate Sales:", sales.duplicated().sum())
print("Duplicate Features:", features.duplicated(subset=['Store', 'Date']).sum())
print("Duplicate Stores:", stores.duplicated(subset=['Store']).sum())

# Check negative sales

print("\nNegative Sales Records:")
print((sales['Weekly_Sales'] < 0).sum())

# Check date range

print("\nSales Date Range:")
print(sales['Date'].min(), "to", sales['Date'].max())

print("\nFeatures Date Range:")
print(features['Date'].min(), "to", features['Date'].max())
```

```python
# Data Cleaning

# 1. Remove duplicate records
sales = sales.drop_duplicates()
features = features.drop_duplicates(subset=['Store', 'Date'])
stores = stores.drop_duplicates(subset=['Store'])

# 2. Fill missing values in Markdown columns
markdown_columns = [
    'MarkDown1',
    'MarkDown2',
    'MarkDown3',
    'MarkDown4',
    'MarkDown5'
]

for col in markdown_columns:
    features[col] = features[col].fillna(0)

# 3. Fill missing CPI and Unemployment values
features['CPI'] = features.groupby('Store')['CPI'].transform(
    lambda x: x.fillna(x.median())
)

features['Unemployment'] = features.groupby('Store')['Unemployment'].transform(
    lambda x: x.fillna(x.median())
)

# 4. Check missing values after cleaning
print("Missing Values After Cleaning:")
print(features.isnull().sum())

print("\nData Cleaning Completed Successfully!")
```

```python
# Data Integration

# Merge Sales with Features
df = pd.merge(
    sales,
    features,
    on=['Store', 'Date'],
    how='left',
    validate='many_to_one'
)

# Merge with Stores
df = pd.merge(
    df,
    stores,
    on='Store',
    how='left',
    validate='many_to_one'
)

# Display merged dataset
print("Merged Dataset Shape:", df.shape)

print("\nMerged Dataset:")
display(df.head())

print("\nMerged Dataset Information:")
df.info()
```

```python
# Feature Engineering

# Extract date information
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.isocalendar().week.astype(int)

# Remove duplicate holiday column
df = df.drop(columns=['IsHoliday_y'])

# Encode Store Type
df = pd.get_dummies(df, columns=['Type'], drop_first=True)

# Display the new dataset
display(df.head())

print("Dataset Shape:", df.shape)
```

```python
negative_sales = df[df['Weekly_Sales'] < 0]

print("Number of Negative Sales Records:")
print(len(negative_sales))

print("\nSample Negative Sales:")
display(negative_sales.head())
```

Insight – Negative Sales

The dataset contains 1,285 records with negative weekly sales. These values may represent product returns or sales adjustments, so they were kept for analysis.

```python
# 1. List Comprehension

sales_list = df['Weekly_Sales'].head(100).tolist()

high_sales_list = [
    sale for sale in sales_list
    if sale > df['Weekly_Sales'].mean()
]

print("High Sales Values:")
print(high_sales_list[:10])


# 2. Lambda Function

calculate_tax = lambda sales: sales * 1.05

print("\nSales after 5% increase:")
print(calculate_tax(1000))
```

```python
# Generator Function

def sales_generator(sales):
    for sale in sales:
        yield sale

gen = sales_generator(sales_list)

print("First 5 Sales Values:")

for i in range(5):
    print(next(gen))
```

```python
# Function with Type Hints

def calculate_average_sales(sales: list) -> float:
    return sum(sales) / len(sales)

average = calculate_average_sales(sales_list)

print("Average Sales:", average)
```

```python
# Decorator Function

def sales_decorator(func):
    def wrapper():
        print("Calculating Sales...")
        func()
        print("Calculation Completed!")
    return wrapper


@sales_decorator
def show_average_sales():
    average = df['Weekly_Sales'].mean()
    print("Average Weekly Sales:", average)


show_average_sales()
```

```python
# Try and Except

try:
    average_sales = df['Weekly_Sales'].mean()
    print("Average Weekly Sales:", average_sales)

except Exception as e:
    print("An error occurred:", e)
```

```python
# Statistical Analysis

print("Sales Mean:", df['Weekly_Sales'].mean())
print("Sales Median:", df['Weekly_Sales'].median())
print("Sales Standard Deviation:", df['Weekly_Sales'].std())
print("Sales Minimum:", df['Weekly_Sales'].min())
print("Sales Maximum:", df['Weekly_Sales'].max())
print("Total Sales:", df['Weekly_Sales'].sum())
```

```python
# Correlation Analysis

correlation = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(12, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Holiday Impact Analysis

normal_sales = df[df['IsHoliday_x'] == False]['Weekly_Sales'].mean()

holiday_week_sales = df[df['IsHoliday_x'] == True]['Weekly_Sales'].mean()

percentage_change = (
    (holiday_week_sales - normal_sales) / normal_sales
) * 100

print("Average Sales During Non-Holiday Weeks:", normal_sales)
print("Average Sales During Holiday Weeks:", holiday_week_sales)
print("Percentage Change:", round(percentage_change, 2), "%")
```

```python
# Train-Test Split (Time-Based)

df = df.sort_values('Date')

split_date = df['Date'].quantile(0.8)

train_data = df[df['Date'] < split_date]
test_data = df[df['Date'] >= split_date]

X_train = train_data.drop(
    columns=['Weekly_Sales', 'Date', 'Type'],
    errors='ignore'
)

y_train = train_data['Weekly_Sales']

X_test = test_data.drop(
    columns=['Weekly_Sales', 'Date', 'Type'],
    errors='ignore'
)

y_test = test_data['Weekly_Sales']

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

```python
# Random Forest Model

from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=20,
    random_state=42,
    n_jobs=2
)

model.fit(X_train, y_train)

print("Random Forest Model Trained Successfully!")
```

```python
# Make Predictions

y_pred = model.predict(X_test)

print("First 10 Predicted Sales Values:")
print(y_pred[:10])
```

```python
# Model Evaluation

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R2 Score:", round(r2, 4))
```

```python
# Actual vs Predicted Sales

comparison = pd.DataFrame({
    'Actual Sales': y_test.values[:10],
    'Predicted Sales': y_pred[:10]
})

display(comparison)
```

```python
# Feature Importance

feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print("Top 10 Important Features:")
display(feature_importance.head(10))
```

```python
# Feature Importance Visualization

top_features = feature_importance.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_features,
    x='Importance',
    y='Feature'
)

plt.title('Top 10 Feature Importance')
plt.xlabel('Importance')
plt.ylabel('Feature')

plt.tight_layout()
plt.show()
```

Insight – Feature Importance

The feature importance analysis shows that Department (Dept) is the most influential factor in predicting weekly sales, followed by Store Size and Store. This indicates that department-level and store-related information play an important role in sales prediction.
